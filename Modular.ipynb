```python
from hierarqcal import (
    Qunitary,
    Qinit,
    Qcycle,
    Qmask,
    plot_circuit,
)

def comparator(bits, symbols=None, state=None):
    v1, v2 = state[bits[0]], state[bits[1]]
    if v1 > v2:
        state[bits[0]], state[bits[1]] = v2, v1
    return state

def modular_addition(nq, tensors):
    addition_mod_n = (
        Qcycle(stride=1, step=1, offset=0, mapping=Qunitary(comparator, n_symbols=0, arity=2), boundary="open")
        + Qmask("*1")
    ) * (nq - 1)
    return addition_mod_n

def modular_subtraction(nq, tensors):
    subtraction_mod_n = (
        Qcycle(stride=1, step=1, offset=0, mapping=Qunitary(comparator, n_symbols=0, arity=2), boundary="open")
        + Qmask("*1")
    ) * (nq - 1)
    return subtraction_mod_n
```

```python
def demonstrate_modular_arithmetic(a, b, N):
    # Number of qubits needed
    nq = max(len(bin(a)), len(bin(b)), len(bin(N))) - 2

    # Initialize quantum state
    tensors = {'a': bin(a)[2:].zfill(nq), 'b': bin(b)[2:].zfill(nq), 'N': bin(N)[2:].zfill(nq)}
    hierq = Qinit(nq, tensors=tensors)

    # Modular Addition
    addition_mod_n = modular_addition(nq, tensors)
    hierq += addition_mod_n
    result_tensor = hierq()
    print("Modular Addition Result:")
    print(result_tensor['b'])

    # Modular Subtraction
    subtraction_mod_n = modular_subtraction(nq, tensors)
    hierq = Qinit(nq, tensors=tensors)  # Reset the quantum state
    hierq += subtraction_mod_n
    result_tensor = hierq()
    print("Modular Subtraction Result:")
    print(result_tensor['b'])
```
